# Chicago Data Analysis Project

## 📊 Project Overview
Complete data analysis using SQL + Python on real Chicago data:
- Chicago Public Schools (safety, attendance)
- Crime Records (533 incidents)
- Census Data (poverty, income, hardship)

## 🔍 Key Findings

### Schools
- Elementary schools average safety score: 49.5
- High schools average safety score: 49.6
- Middle schools average safety score: 48.0

### Crime
- Total crimes analyzed: 533
- Most crime-prone community: Area 25 (43 crimes)
- School-related crimes: Battery, Narcotics, Assault, Criminal Trespass

### Poverty
- Highest poverty areas: Riverdale (56.5%), Fuller Park (51.2%)
- 4 communities with per capita income < $11,000

## 🛠️ Technologies
- Python (Pandas, SQLite3)
- SQL (Aggregations, JOINs, Subqueries)
- Jupyter Notebook

## 📈 Sample Queries
```sql
-- Average safety score by school type
SELECT [Elementary, Middle, or High School], AVG(SAFETY_SCORE) 
FROM CHICAGO_PUBLIC_SCHOOLS 
GROUP BY [Elementary, Middle, or High School];

-- Most crime-prone area
SELECT COMMUNITY_AREA_NUMBER, COUNT(*) as crime_count 
FROM CHICAGO_CRIME_DATA 
GROUP BY COMMUNITY_AREA_NUMBER 
ORDER BY crime_count DESC LIMIT 1;

````python
#List all kidnapping crimes involving a child?
%sql SELECT CASE_NUMBER FROM CHICAGO_CRIME_DATA WHERE PRIMARY_TYPE = 'KIDNAPPING' AND DESCRIPTION LIKE '%CHILD%';
````

````python
#List the kind of crimes that were recorded at schools. (No repetitions)
%sql SELECT DISTINCT PRIMARY_TYPE FROM CHICAGO_CRIME_DATA WHERE LOCATION_DESCRIPTION LIKE '%SCHOOL%';
````
